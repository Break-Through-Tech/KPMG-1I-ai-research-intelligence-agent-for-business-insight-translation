```python
import arxiv

client = arxiv.Client()
search = arxiv.Search(
    query="LLM routing",
    max_results=10,
    sort_by=arxiv.SortCriterion.SubmittedDate
)

results = list(client.results(search))

for paper in results:
    print(paper.title)
    print(paper.published)
    print(paper.entry_id)
    print("---")
```
